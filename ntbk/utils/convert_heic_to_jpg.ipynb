```python
from pathlib import Path
import pillow_heif
from PIL import Image

pillow_heif.register_heif_opener()

def convert_dataset_heic_to_jpg(source_dir, output_dir):
    source = Path(source_dir)
    
    for class_folder in source.iterdir():
        if not class_folder.is_dir() or class_folder.name == "converted":
            continue
        
        output_class_dir = Path(output_dir) / class_folder.name
        output_class_dir.mkdir(parents=True, exist_ok=True)
        
        heif_files = list(class_folder.glob("*.HEIC"))
        print(f"\n[{class_folder.name}] Found {len(heif_files)} HEIC files")
        
        for i, heic_file in enumerate(heif_files):
            jpg_path = output_class_dir / (heic_file.stem + ".jpg")
            Image.open(heic_file).save(jpg_path, "JPEG")
            print(f"  [{i+1}/{len(heif_files)}] {heic_file.name} -> {jpg_path.name}")

source_dir = r"../dataset"
output_dir = r"../dataset/converted"
convert_dataset_heic_to_jpg(source_dir, output_dir)
```
