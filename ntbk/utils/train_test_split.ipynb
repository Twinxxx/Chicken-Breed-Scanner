```python
from pathlib import Path
import shutil
import random

def train_test_split_dataset(source_dir, output_dir, train_ratio=0.7, seed=42):
    random.seed(seed)
    source = Path(source_dir)
    output = Path(output_dir)

    for class_folder in source.iterdir():
        if not class_folder.is_dir():
            continue

        images = list(class_folder.glob("*.jpg"))
        random.shuffle(images)

        split_idx = int(len(images) * train_ratio)
        train_files = images[:split_idx]
        test_files = images[split_idx:]

        # Create train/test subfolders per class
        train_dir = output / "train" / class_folder.name
        test_dir = output / "test" / class_folder.name
        train_dir.mkdir(parents=True, exist_ok=True)
        test_dir.mkdir(parents=True, exist_ok=True)

        for f in train_files:
            shutil.copy(f, train_dir / f.name)
        for f in test_files:
            shutil.copy(f, test_dir / f.name)

        print(f"[{class_folder.name}] {len(images)} total -> {len(train_files)} train / {len(test_files)} test")

source_dir = r"../dataset/converted"
output_dir = r"../dataset/splitted"
train_test_split_dataset(source_dir, output_dir)
```
